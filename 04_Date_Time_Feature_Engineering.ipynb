```python
import pandas as pd
import numpy as np
from datetime import datetime

print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Purchase_Date": [
        "2024-01-15",
        "2024-03-22",
        "2024-06-10",
        "2024-09-18",
        "2024-12-25"
    ],
    "Customer_Name": [
        "Arun", "Rahul", "Priya", "Kiran", "Sneha"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])

print("Date column converted successfully!")
print(df.dtypes)

display(df)
```

```python
df["Year"] = df["Purchase_Date"].dt.year

display(df[["Purchase_Date", "Year"]])
```

```python
df["Month"] = df["Purchase_Date"].dt.month

display(df[["Purchase_Date", "Month"]])
```

```python
df["Day"] = df["Purchase_Date"].dt.day

display(df[["Purchase_Date", "Day"]])
```

```python
df["Day_of_Week"] = df["Purchase_Date"].dt.day_name()

display(df[["Purchase_Date", "Day_of_Week"]])
```

```python
df["Week_Number"] = df["Purchase_Date"].dt.isocalendar().week

display(df[["Purchase_Date", "Week_Number"]])
```

```python
df["Quarter"] = df["Purchase_Date"].dt.quarter

display(df[["Purchase_Date", "Quarter"]])
```

```python
df["Weekend_Indicator"] = (
    df["Purchase_Date"].dt.dayofweek >= 5
).astype(int)

display(df[["Purchase_Date", "Weekend_Indicator"]])
```

```python
df["Month_Start"] = (
    df["Purchase_Date"].dt.is_month_start
).astype(int)

display(df[["Purchase_Date", "Month_Start"]])
```

```python
df["Month_End"] = (
    df["Purchase_Date"].dt.is_month_end
).astype(int)

display(df[["Purchase_Date", "Month_End"]])
```

```python
reference_date = pd.Timestamp("2025-01-01")

df["Days_Since_Event"] = (
    reference_date - df["Purchase_Date"]
).dt.days

display(df[["Purchase_Date", "Days_Since_Event"]])
```

```python
print("Dataset with Date & Time Features:")
display(df)
```

```python
df["Previous_Purchase_Date"] = pd.to_datetime([
    "2024-01-01",
    "2024-03-01",
    "2024-05-01",
    "2024-07-01",
    "2024-09-01"
])

df["Time_Difference"] = (
    df["Purchase_Date"] - df["Previous_Purchase_Date"]
).dt.days

display(df[[
    "Previous_Purchase_Date",
    "Purchase_Date",
    "Time_Difference"
]])
```

```python
df["Date_of_Birth"] = pd.to_datetime([
    "1995-05-10",
    "1998-08-15",
    "1992-03-20",
    "2000-11-25",
    "1996-07-12"
])

reference_date = pd.Timestamp("2025-01-01")

df["Customer_Age"] = (
    reference_date.year - df["Date_of_Birth"].dt.year
    - (
        (reference_date.month < df["Date_of_Birth"].dt.month) |
        (
            (reference_date.month == df["Date_of_Birth"].dt.month) &
            (reference_date.day < df["Date_of_Birth"].dt.day)
        )
    ).astype(int)
)

display(df[["Customer_Name", "Date_of_Birth", "Customer_Age"]])
```

```python
df["Join_Date"] = pd.to_datetime([
    "2022-01-10",
    "2022-06-15",
    "2023-01-20",
    "2023-05-25",
    "2024-01-01"
])

df["Customer_Tenure_Days"] = (
    reference_date - df["Join_Date"]
).dt.days

display(df[["Customer_Name", "Join_Date", "Customer_Tenure_Days"]])
```

```python
df["Last_Purchase_Date"] = pd.to_datetime([
    "2024-12-20",
    "2024-11-15",
    "2024-10-10",
    "2024-12-01",
    "2024-12-25"
])

df["Recency"] = (
    reference_date - df["Last_Purchase_Date"]
).dt.days

display(df[["Customer_Name", "Last_Purchase_Date", "Recency"]])
```

```python
purchase_counts = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Frequency": [5, 3, 8, 2, 6]
})

df = df.merge(purchase_counts, on="Customer_ID", how="left")

display(df[["Customer_ID", "Customer_Name", "Frequency"]])
```

```python
df["Purchase_Month"] = df["Purchase_Date"].dt.month

def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df["Season"] = df["Purchase_Month"].apply(get_season)

display(df[["Purchase_Date", "Purchase_Month", "Season"]])
```

```python
seasonal_frequency = df["Season"].value_counts()

print("Seasonal Purchase Frequency:")
display(seasonal_frequency.to_frame(name="Purchase_Count"))
```

```python
monthly_frequency = df["Purchase_Month"].value_counts().sort_index()

print("Monthly Purchase Frequency:")
display(monthly_frequency.to_frame(name="Purchase_Count"))
```

```python
monthly_frequency = df["Purchase_Month"].value_counts().sort_index()

print("Monthly Purchase Frequency:")
display(monthly_frequency.to_frame(name="Purchase_Count"))
```

```python
print("Final Date & Time Feature Engineering Dataset:")
display(df)

print("\nDataset Shape:", df.shape)

print("\nGenerated Features:")
print(df.columns.tolist())
```
