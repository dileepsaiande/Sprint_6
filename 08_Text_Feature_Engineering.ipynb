```python
import pandas as pd
import numpy as np
import re

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Review": [
        "This product is excellent! I love it 100 times.",
        "Good quality and fast delivery.",
        "The product is average, but useful.",
        "Excellent service! Highly recommended.",
        "Bad experience. The product was not good."
    ]
}

df = pd.DataFrame(data)

print("Original Text Dataset:")
display(df)
```

```python
df["Text_Length"] = df["Review"].str.len()

display(df[["Review", "Text_Length"]])
```

```python
df["Character_Count"] = df["Review"].apply(len)

display(df[["Review", "Character_Count"]])
```

```python
df["Word_Count"] = df["Review"].apply(
    lambda text: len(text.split())
)

display(df[["Review", "Word_Count"]])
```

```python
df["Sentence_Count"] = df["Review"].apply(
    lambda text: len([
        sentence for sentence in re.split(r'[.!?]+', text)
        if sentence.strip()
    ])
)

display(df[["Review", "Sentence_Count"]])
```

```python
df["Digit_Count"] = df["Review"].apply(
    lambda text: sum(character.isdigit() for character in text)
)

display(df[["Review", "Digit_Count"]])
```

```python
df["Special_Character_Count"] = df["Review"].apply(
    lambda text: sum(
        not character.isalnum() and not character.isspace()
        for character in text
    )
)

display(df[["Review", "Special_Character_Count"]])
```

```python
df["Uppercase_Count"] = df["Review"].apply(
    lambda text: sum(character.isupper() for character in text)
)

display(df[["Review", "Uppercase_Count"]])
```

```python
df["Lowercase_Count"] = df["Review"].apply(
    lambda text: sum(character.islower() for character in text)
)

display(df[["Review", "Lowercase_Count"]])
```

```python
keywords = ["excellent", "good", "bad", "product", "delivery"]

def count_keywords(text):
    words = re.findall(r"\b\w+\b", text.lower())
    return sum(words.count(keyword) for keyword in keywords)

df["Keyword_Count"] = df["Review"].apply(count_keywords)

display(df[["Review", "Keyword_Count"]])
```

```python
vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(df["Review"])

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("Bag of Words Features:")
display(bow_df)
```

```python
print("Dataset with Text Features:")
display(df)

print("Dataset Shape:", df.shape)
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(df["Review"])

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("TF-IDF Features:")
display(tfidf_df)
```

```python
ngram_vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

ngram_matrix = ngram_vectorizer.fit_transform(df["Review"])

ngram_df = pd.DataFrame(
    ngram_matrix.toarray(),
    columns=ngram_vectorizer.get_feature_names_out()
)

print("Unigram and Bigram Features:")
display(ngram_df)
```

```python
print("Bag of Words Shape:", bow_df.shape)
print("TF-IDF Shape:", tfidf_df.shape)
print("N-Grams Shape:", ngram_df.shape)
```

```python
print("Final Text Feature Dataset:")
display(df)

print("Dataset Shape:", df.shape)
```
