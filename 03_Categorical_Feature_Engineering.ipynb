## 3. Importance

Many machine learning algorithms require categorical data to be represented numerically.


## 4. Categorical Variables

A categorical variable represents a group, class, or label.

Examples:
- Location: Chennai, Hyderabad, Bangalore
- Property_Type: Apartment, Villa
- Furnished: Yes, No

Categorical variables can be nominal, ordinal, or binary.

When to use:
Categorical variables are useful when the information represents distinct groups rather than numerical measurements.


```python

import pandas as pd
import numpy as np

# Create the sample property dataset
data = {
    "Property_ID": [101, 102, 103, 104, 105],
    "Location": ["Chennai", "Hyderabad", "Bangalore", "Chennai", "Hyderabad"],
    "Property_Type": ["Apartment", "Villa", "Apartment", "Villa", "Apartment"],
    "Condition": ["Good", "Excellent", "Average", "Good", "Excellent"],
    "Furnished": ["Yes", "No", "Yes", "No", "Yes"],
    "Price": [3000000, 4500000, 6000000, 3600000, 7500000]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
display(df)
```


## 5. Nominal Variables

Nominal variables represent categories without a meaningful ranking or order.

Examples:
- Location: Chennai, Hyderabad, Bangalore
- Property_Type: Apartment, Villa
- Color: Red, Blue, Green

When to use:
Nominal variables are useful for representing categories where one value is not inherently greater or better than another.

One-Hot Encoding is commonly used for nominal variables.


```python

nominal_columns = ["Location", "Property_Type"]

for column in nominal_columns:
    print(f"\n{column} categories:")
    print(df[column].unique())
```


## 6. Ordinal Variables

Ordinal variables represent categories with a meaningful order.

Examples:
- Property condition: Poor, Average, Good, Excellent
- Education level: School, Bachelor's, Master's, Doctorate
- Satisfaction: Low, Medium, High

When to use:
Ordinal variables are useful when categories have a meaningful ranking.

Ordinal Encoding is commonly used because it preserves the order between categories.


```python

condition_order = {
    "Poor": 1,
    "Average": 2,
    "Good": 3,
    "Excellent": 4
}

df["Condition_Encoded"] = df["Condition"].map(condition_order)

display(df[["Condition", "Condition_Encoded"]])
```


## 7. Binary Variables

Binary variables represent exactly two possible categories.

Examples:
- Furnished: Yes, No
- Loan_Status: Approved, Rejected
- Property_Available: True, False

When to use:
Binary variables are useful when a feature represents two mutually exclusive states.

They can often be encoded using 0 and 1.


```python

df["Furnished_Encoded"] = df["Furnished"].map({
    "No": 0,
    "Yes": 1
})

display(df[["Furnished", "Furnished_Encoded"]])
```


## 8. One-Hot Encoding

One-Hot Encoding converts each category into a separate binary column.

Example:
Location = Chennai, Hyderabad, Bangalore

The resulting columns are:
- Location_Chennai
- Location_Hyderabad
- Location_Bangalore

When to use:
- For nominal variables without a natural order.
- When the number of unique categories is manageable.
- When category identity is important.

Advantages:
- Does not introduce an artificial ranking.
- Is easy to interpret.

Limitation:
- Can create many columns when a variable has high cardinality.


```python


# Apply One-Hot Encoding to Location
df_onehot = pd.get_dummies(
    df,
    columns=["Location"],
    dtype=int
)

display(df_onehot.filter(regex="Location_"))
```


## 9. Ordinal Encoding

Ordinal Encoding converts ordered categories into numerical values while preserving their ranking.

Example:
- Poor = 1
- Average = 2
- Good = 3
- Excellent = 4

When to use:
- For categories with a meaningful order.
- When the ranking between categories is important.

Advantages:
- Preserves the order of categories.
- Uses fewer columns than One-Hot Encoding.

Limitation:
- The numerical distances between categories may not represent real differences.


```python

from sklearn.preprocessing import OrdinalEncoder

condition_order = [
    ["Poor", "Average", "Good", "Excellent"]
]

encoder = OrdinalEncoder(
    categories=condition_order
)

df["Condition_Ordinal"] = encoder.fit_transform(
    df[["Condition"]]
).ravel()

display(df[["Condition", "Condition_Ordinal"]])
```


## 10. Label Encoding

Label Encoding converts categorical labels into integer values.

Example:
- Apartment = 0
- Villa = 1

When to use:
- For categorical target labels in classification.
- When a model requires numerical class labels.

Advantages:
- Simple and memory-efficient.
- Useful for representing classification targets.

Limitation:
- Integer labels can introduce an artificial order if used for nominal input features.

For unordered input features, One-Hot Encoding is often more appropriate.


```python

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df["Property_Type_Encoded"] = label_encoder.fit_transform(
    df["Property_Type"]
)

print("Category mapping:")
for category, encoded_value in zip(
    label_encoder.classes_,
    label_encoder.transform(label_encoder.classes_)
):
    print(category, "=", encoded_value)

display(df[["Property_Type", "Property_Type_Encoded"]])
```


## 11. Frequency Encoding

Frequency Encoding replaces each category with its frequency or proportion in the dataset.

Example:
If Chennai appears 3 times out of 5 records, its relative frequency is 0.60.

When to use:
- For categorical variables with many unique categories.
- When category frequency may contain useful information.
- When One-Hot Encoding would create too many columns.

Limitation:
Different categories can have the same frequency, so their identities may be lost.

Calculate frequency mappings using training data only.


```python

# Calculate relative frequency using Location
location_frequency = df["Location"].value_counts(normalize=True)

df["Location_Frequency"] = (
    df["Location"].map(location_frequency)
)

display(df[["Location", "Location_Frequency"]])
```


## 12. Target Encoding

Target Encoding replaces each category with a statistic derived from the target variable.

Example:
Each location can be represented by its average property price.

When to use:
- For high-cardinality categorical features.
- When the category's relationship with the target is useful.
- When One-Hot Encoding creates too many features.

Advantages:
- Can capture relationships between categories and the target.
- Produces a compact numerical representation.

Limitation:
- Can cause target leakage and overfitting if implemented incorrectly.

Use cross-fitting or out-of-fold target encoding during training. Never calculate target encodings using the validation or test targets.


```python


# Demonstration only: calculate average price by location
location_target_mean = df.groupby("Location")["Price"].mean()

df["Location_Target_Encoded"] = (
    df["Location"].map(location_target_mean)
)

display(df[["Location", "Price", "Location_Target_Encoded"]])
```


## 13. Rare Category Grouping

Rare category grouping combines infrequent categories into a single category.

Example:
If a location appears in fewer than 5% of the training records, it can be grouped into "Other".

When to use:
- When a categorical variable has many unique values.
- When some categories appear very infrequently.
- When reducing the number of encoded features is useful.

Benefits:
- Reduces dimensionality.
- Can improve model stability.
- Helps manage rare categories.

The grouping threshold should be determined using training data only.


```python

# Create a sample dataset with rare categories
locations = pd.Series([
    "Chennai", "Chennai", "Chennai",
    "Hyderabad", "Hyderabad",
    "Bangalore", "Mumbai", "Delhi"
])

# Calculate relative frequencies
location_frequency = locations.value_counts(normalize=True)

# Identify categories appearing in less than 20% of records
rare_categories = location_frequency[
    location_frequency < 0.20
].index

# Group rare categories
locations_grouped = locations.replace(
    rare_categories, "Other"
)

display(pd.DataFrame({
    "Original_Location": locations,
    "Grouped_Location": locations_grouped
}))
```


## 14. Category Cardinality

Category cardinality is the number of unique values in a categorical feature.

Examples:
- Property_Type may have low cardinality.
- Location may have moderate cardinality.
- Customer_ID may have very high cardinality.

When to use:
Cardinality helps determine which encoding technique is appropriate.

Low cardinality:
One-Hot Encoding is often suitable.

High cardinality:
Frequency Encoding, Target Encoding, or rare category grouping may be more appropriate.

High-cardinality features can increase memory usage and create too many columns when One-Hot Encoding is used.


```python

import pandas as pd

# Create a sample dataset
data = {
    "Location": ["Chennai", "Hyderabad", "Bangalore", "Chennai", "Hyderabad"],
    "Property_Type": ["Apartment", "Villa", "Apartment", "Villa", "Apartment"],
    "Condition": ["Good", "Excellent", "Average", "Good", "Excellent"],
    "Furnished": ["Yes", "No", "Yes", "No", "Yes"]
}

# Calculate unique categories
cardinality = {}

for column, values in data.items():
    cardinality[column] = len(set(values))

# Display results
result = pd.DataFrame.from_dict(
    cardinality,
    orient="index",
    columns=["Unique_Categories"]
)

print("Category Cardinality:")
display(result)
```


## 15. Handling Unknown Categories

Unknown categories are categories encountered during prediction that were not present during training.

Example:
A model trained on Chennai and Hyderabad may receive a property from Pune.

When to use:
When categorical features can contain new values in production.

Common strategies:
- Map unknown values to an "Other" category.
- Use encoders that support unknown categories.
- Use a fallback category or default representation.

Handling unknown categories prevents errors when the model receives new data.


```python

from sklearn.preprocessing import OneHotEncoder

# Training data
X_train = pd.DataFrame({
    "Location": ["Chennai", "Hyderabad", "Bangalore"]
})

# Create an encoder that ignores unknown categories
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(X_train)

# New data contains an unknown category
X_new = pd.DataFrame({
    "Location": ["Pune"]
})

encoded_new = encoder.transform(X_new)

encoded_df = pd.DataFrame(
    encoded_new,
    columns=encoder.get_feature_names_out(["Location"])
)

display(encoded_df)
```


## 16. Combining Categories

Combining categories merges related categorical values into a single category.

Example:
- Apartment and Flat → Apartment
- Villa and Independent House → Independent House

When to use:
- When categories have similar meanings.
- When reducing the number of categories is useful.
- When domain knowledge supports grouping.

Combining categories can simplify a dataset and make patterns easier to identify.

Category combinations should be based on meaningful relationships, not arbitrary grouping.


```python

# Sample property types
df["Property_Type"] = [
    "Apartment",
    "Villa",
    "Flat",
    "Independent House",
    "Apartment"
]

# Combine related categories
category_mapping = {
    "Apartment": "Apartment",
    "Flat": "Apartment",
    "Villa": "Independent House",
    "Independent House": "Independent House"
}

df["Combined_Property_Type"] = (
    df["Property_Type"].map(category_mapping)
)

display(df[["Property_Type", "Combined_Property_Type"]])
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

# Sample dataset
df = pd.DataFrame({
    "Location": ["Chennai", "Hyderabad", "Bangalore", "Chennai", "Hyderabad"],
    "Condition": ["Good", "Excellent", "Average", "Good", "Excellent"],
    "Property_Type": ["Apartment", "Villa", "Apartment", "Villa", "Apartment"]
})

# One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=["Location"], dtype=int)

# Ordinal Encoding
condition_encoder = OrdinalEncoder(
    categories=[["Average", "Good", "Excellent"]]
)
df_encoded["Condition_Encoded"] = condition_encoder.fit_transform(
    df[["Condition"]]
).astype(int)

# Label Encoding
property_encoder = LabelEncoder()
df_encoded["Property_Type_Encoded"] = property_encoder.fit_transform(
    df["Property_Type"]
)

# Category cardinality
categorical_columns = df.select_dtypes(include=["object", "category"]).columns
print("Categorical columns:", categorical_columns.tolist())
print("\nCategory cardinality:")
display(df[categorical_columns].nunique())

print("\nFinal encoded dataset:")
display(df_encoded)
```
