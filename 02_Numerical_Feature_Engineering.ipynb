
# Notebook 2 – Numerical Feature Engineering

## 1. Introduction

Numerical feature engineering involves creating and transforming numerical features to help machine learning models learn meaningful patterns.

## 2. Mathematical Transformations

Mathematical transformations use arithmetic and mathematical operations to create new features from existing numerical data.

Common operations include addition, subtraction, multiplication, division, ratios, and percentages.

These transformations help represent relationships between variables and provide useful information to machine learning models.

## 3. When and Why to Use

- Use mathematical transformations when relationships between variables are meaningful.
- Create additional information from existing numerical features.
- Help models identify patterns that may not be obvious in raw data.
- Improve model interpretability by representing meaningful quantities.


```python

import pandas as pd
import numpy as np

# Sample property dataset
data = {
    "Area_sqft": [1000, 1500, 2000, 2500, 3000],
    "Bedrooms": [2, 3, 4, 5, 6],
    "Bathrooms": [1, 2, 3, 4, 5],
    "Price": [3000000, 4500000, 6000000, 10000000, 12000000]
}

df = pd.DataFrame(data)

display(df)
```


## 4. Addition

Addition combines numerical features to create a new feature.

Example:
Total_Rooms = Bedrooms + Bathrooms

It is useful when the combined quantity provides meaningful information for the prediction task.


```python

df["Total_Rooms"] = df["Bedrooms"] + df["Bathrooms"]

display(df[["Bedrooms", "Bathrooms", "Total_Rooms"]])
```


## 5. Subtraction

Subtraction calculates the difference between numerical variables.

Example:
Property_Age = Current_Year - Year_Built

It is useful for identifying differences, remaining balances, and time-based quantities.


```python

df["Year_Built"] = [2010, 2015, 2020, 2012, 2022]

df["Property_Age"] = 2026 - df["Year_Built"]

display(df[["Year_Built", "Property_Age"]])
```


## 6. Multiplication

Multiplication creates a new feature by multiplying two numerical variables.

Example:
Estimated_Price = Area_sqft × Price_per_sqft

It is useful when two variables have a meaningful multiplicative relationship.


```python

df["Price_per_sqft"] = [3000, 3000, 3000, 4000, 4000]

df["Estimated_Price"] = (
    df["Area_sqft"] * df["Price_per_sqft"]
)

display(df[["Area_sqft", "Price_per_sqft", "Estimated_Price"]])
```


## 7. Division

Division creates a new feature by dividing one numerical variable by another.

Example:
Price_per_sqft = Price / Area_sqft

It is useful for calculating rates, unit values, and relative quantities.

Always check for zero values in the denominator to avoid division errors.


```python

df["Price_per_sqft"] = df["Price"] / df["Area_sqft"]

display(df[["Price", "Area_sqft", "Price_per_sqft"]])
```


## 8. Ratios

A ratio represents the relationship between two numerical variables.

Example:
Bathroom_Bedroom_Ratio = Bathrooms / Bedrooms

When to use:
- To compare quantities relative to one another.
- To identify proportional relationships.
- To create features that are less dependent on the absolute size of a property.

Ratios are useful when the relative relationship between variables matters more than their individual values.

Always handle zero values in the denominator.


```python

df["Bathroom_Bedroom_Ratio"] = (
    df["Bathrooms"] / df["Bedrooms"]
)

display(df[["Bedrooms", "Bathrooms", "Bathroom_Bedroom_Ratio"]])
```


## 9. Percentages

A percentage expresses a numerical value as a proportion of 100.

Formula:
Percentage = (Part / Total) × 100

Example:
Discount_Percentage = (Original_Price - Final_Price) / Original_Price × 100

When to use:
- To calculate discounts or growth rates.
- To compare relative changes.
- To represent a quantity as a percentage of a total.

Percentages make comparisons easier when the original quantities differ in size.


```python

df["Original_Price"] = [3500000, 5000000, 7000000, 12000000, 15000000]

df["Discount_Percentage"] = (
    (df["Original_Price"] - df["Price"])
    / df["Original_Price"]
) * 100

display(df[["Original_Price", "Price", "Discount_Percentage"]].round(2))
```


## 10. Differences

A difference measures the change between two numerical values.

Formula:
Difference = Current_Value - Previous_Value

Example:
Price_Change = Current_Price - Previous_Price

When to use:
- To measure increases or decreases.
- To compare current and historical values.
- To identify changes over time.

A positive difference indicates an increase, while a negative difference indicates a decrease.


```python

df["Previous_Price"] = [2800000, 4200000, 5500000, 9500000, 11000000]

df["Price_Difference"] = (
    df["Price"] - df["Previous_Price"]
)

display(df[["Previous_Price", "Price", "Price_Difference"]])
```


## 11. Aggregations

Aggregation combines multiple values into a summary statistic.

Common aggregation functions:
- Sum
- Mean
- Minimum
- Maximum
- Count

Example:
Average_Price = Mean of property prices in a location.

When to use:
- To summarize groups of observations.
- To calculate average values.
- To identify patterns across categories.

Aggregations are useful for creating group-level features.


```python

df["Location"] = ["Chennai", "Hyderabad", "Chennai", "Hyderabad", "Chennai"]

# Calculate average property price by location
location_summary = df.groupby("Location").agg(
    Average_Price=("Price", "mean"),
    Total_Properties=("Price", "count"),
    Maximum_Price=("Price", "max")
).reset_index()

display(location_summary)
```


## 12. Log Transformation

Log transformation applies a logarithmic function to numerical values.

Formula:
Log_Transformed = log(1 + x)

When to use:
- To reduce right skewness.
- To reduce the influence of extremely large values.
- To help models learn relationships in highly skewed numerical data.

The log1p function is useful because it calculates log(1 + x) and can handle zero values.

Log transformation is not suitable for negative values without an appropriate adjustment.


```python

# Apply log transformation to property prices
df["Log_Price"] = np.log1p(df["Price"])

display(df[["Price", "Log_Price"]].round(3))
```


## 13. Square Root Transformation

Square root transformation applies the square root function to a numerical feature.

Formula:
Transformed_Value = √x

When to use:
- To reduce right skewness in numerical data.
- To reduce the effect of large values.
- To transform non-negative numerical features.

It is commonly useful for count-based features, such as the number of nearby hospitals or schools.

Square root transformation requires non-negative values.


```python

df["Sqrt_Price"] = np.sqrt(df["Price"])

display(df[["Price", "Sqrt_Price"]].round(2))
```


## 14. Power Transformation

Power transformation changes numerical values using a mathematical power function.

Formula:
Transformed_Value = x^p

Where p is the selected power.

When to use:
- To change the distribution of numerical features.
- To reduce skewness.
- To help models learn relationships in numerical data.

The Yeo-Johnson transformation can handle both positive and negative values. The Box-Cox transformation requires strictly positive values.

The appropriate transformation depends on the distribution of the data.


```python

from sklearn.preprocessing import PowerTransformer

# Create a positive numerical feature
X = df[["Area_sqft"]]

# Apply Yeo-Johnson power transformation
power_transformer = PowerTransformer(method="yeo-johnson")

df["Power_Transformed_Area"] = (
    power_transformer.fit_transform(X).ravel()
)

display(df[["Area_sqft", "Power_Transformed_Area"]].round(3))
```


## 15. Absolute Difference

Absolute difference calculates the non-negative distance between two numerical values.

Formula:
Absolute_Difference = |A - B|

When to use:
- To measure the distance between two values.
- To compare current and previous measurements.
- To measure the difference between actual and expected values.

Absolute difference ignores whether the change is positive or negative.


```python

df["Absolute_Price_Difference"] = (
    df["Price"] - df["Previous_Price"]
).abs()

display(df[["Price", "Previous_Price", "Absolute_Price_Difference"]])
```


## 16. Relative Difference

Relative difference measures the difference between two values relative to a reference value.

Formula:
Relative_Difference = (Current_Value - Previous_Value) / Previous_Value

When to use:
- To compare changes across different scales.
- To measure proportional increases or decreases.
- To compare growth between different properties or groups.

Multiply the result by 100 to express it as a percentage.

The reference value must not be zero.


```python

df["Relative_Price_Difference"] = (
    (df["Price"] - df["Previous_Price"])
    / df["Previous_Price"]
)

display(df[["Price", "Previous_Price", "Relative_Price_Difference"]].round(3))
```


## 17. Rate Features

Rate features express a quantity per unit of time or another meaningful unit.

Formula:
Rate = Change in Quantity / Time Interval

Example:
Price_Growth_Per_Year = (Current_Price - Previous_Price) / Number_of_Years

When to use:
- To measure growth over time.
- To compare the speed of change between observations.
- To calculate quantities such as price growth per year or sales per day.

Always ensure that the time interval is meaningful and non-zero.


```python

df["Years_Elapsed"] = [2, 3, 4, 5, 6]

df["Price_Growth_Per_Year"] = (
    (df["Price"] - df["Previous_Price"])
    / df["Years_Elapsed"]
)

display(df[["Price", "Previous_Price", "Years_Elapsed", "Price_Growth_Per_Year"]])
```


## 18. Normalized Features

Normalization scales numerical features to a specified range, commonly between 0 and 1.

Min-Max Normalization formula:

Normalized_Value = (x - Minimum) / (Maximum - Minimum)

When to use:
- When numerical features have different scales.
- When a machine learning algorithm is sensitive to feature magnitude.
- When a common numerical range is useful for model training.

Normalization can improve the performance of algorithms such as K-Nearest Neighbors and neural networks.

It is not always necessary for tree-based models such as Random Forest.

Important: Fit the normalization parameters using training data only to prevent data leakage.


```python

from sklearn.preprocessing import MinMaxScaler

# Select numerical features
features = df[["Area_sqft", "Bedrooms", "Bathrooms"]]

# Apply Min-Max Normalization
scaler = MinMaxScaler()

normalized_values = scaler.fit_transform(features)

normalized_df = pd.DataFrame(
    normalized_values,
    columns=["Normalized_Area", "Normalized_Bedrooms", "Normalized_Bathrooms"]
)

display(normalized_df.round(2))
```
