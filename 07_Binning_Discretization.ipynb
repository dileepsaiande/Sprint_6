```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": range(101, 111),
    "Age": [18, 22, 25, 29, 35, 40, 45, 50, 58, 65],
    "Income": [
        15000, 20000, 25000, 30000, 40000,
        50000, 60000, 75000, 90000, 120000
    ],
    "Risk_Score": [15, 25, 35, 45, 55, 60, 70, 75, 85, 95]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
df["Age_Category"] = pd.cut(
    df["Age"],
    bins=[0, 25, 45, 100],
    labels=["Young", "Adult", "Senior"]
)

display(df[["Age", "Age_Category"]])
```

```python
df["Income_Category"] = pd.cut(
    df["Income"],
    bins=[0, 30000, 70000, 150000],
    labels=["Low", "Medium", "High"]
)

display(df[["Income", "Income_Category"]])
```

```python
df["Equal_Width_Bin"] = pd.cut(
    df["Age"],
    bins=4,
    labels=["Bin_1", "Bin_2", "Bin_3", "Bin_4"]
)

display(df[["Age", "Equal_Width_Bin"]])
```

```python
df["Equal_Frequency_Bin"] = pd.qcut(
    df["Income"],
    q=4,
    labels=["Low", "Medium", "High", "Very_High"]
)

display(df[["Income", "Equal_Frequency_Bin"]])
```

```python
df["Quantile_Bin"] = pd.qcut(
    df["Risk_Score"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

display(df[["Risk_Score", "Quantile_Bin"]])
```

```python
df["Quantile_Bin"] = pd.qcut(
    df["Risk_Score"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

display(df[["Risk_Score", "Quantile_Bin"]])
```

```python
df["Domain_Income_Bin"] = pd.cut(
    df["Income"],
    bins=[0, 25000, 50000, 100000, float("inf")],
    labels=["Low", "Lower_Middle", "Upper_Middle", "High"]
)

display(df[["Income", "Domain_Income_Bin"]])
```

```python
df["Risk_Category"] = pd.cut(
    df["Risk_Score"],
    bins=[0, 30, 60, 100],
    labels=["Low_Risk", "Medium_Risk", "High_Risk"]
)

display(df[["Risk_Score", "Risk_Category"]])
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=[
        "Teenager",
        "Young_Adult",
        "Adult",
        "Middle_Aged",
        "Senior"
    ],
    include_lowest=True
)

display(df[["Age", "Age_Group"]])
```

```python
df["Income_Group"] = pd.cut(
    df["Income"],
    bins=[0, 30000, 60000, 90000, float("inf")],
    labels=["Low", "Medium", "High", "Very_High"],
    include_lowest=True
)

display(df[["Income", "Income_Group"]])
```

```python
df["Risk_Level"] = pd.cut(
    df["Risk_Score"],
    bins=[0, 40, 70, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

display(df[["Customer_ID", "Risk_Score", "Risk_Level"]])
```

```python
print("Final Dataset with Binning Features:")
display(df)

print("Dataset Shape:", df.shape)
```
