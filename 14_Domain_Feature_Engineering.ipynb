```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],

    "Annual_Income": [300000, 450000, 600000, 800000,
                      500000, 900000, 350000, 700000,
                      1000000, 550000],

    "Loan_Amount": [100000, 200000, 250000, 300000,
                    150000, 400000, 120000, 280000,
                    450000, 180000],

    "Monthly_Debt": [5000, 10000, 12000, 15000, 8000,
                     20000, 6000, 14000, 25000, 9000],

    "Credit_Limit": [50000, 80000, 100000, 150000,
                     70000, 200000, 60000, 120000,
                     250000, 90000],

    "Credit_Used": [20000, 50000, 30000, 120000,
                    25000, 150000, 45000, 60000,
                    200000, 40000],

    "Employment_Years": [2, 5, 8, 10, 3,
                         12, 1, 7, 15, 4],

    "Monthly_Savings": [3000, 5000, 10000, 12000, 4000,
                        15000, 2000, 9000, 20000, 6000],

    "Previous_Loans": [1, 2, 3, 2, 1, 4, 0, 3, 5, 2],

    "Missed_Payments": [0, 2, 0, 1, 3, 0, 4, 1, 0, 2],

    "Loan_Default": [0, 1, 0, 0, 1, 0, 1, 0, 0, 1]
}

df = pd.DataFrame(data)

print("Original Banking Dataset:")
display(df)

print("Dataset Shape:", df.shape)
```

```python
df["Debt_to_Income_Ratio"] = (
    df["Monthly_Debt"] /
    (df["Annual_Income"] / 12)
)

display(df[[
    "Annual_Income",
    "Monthly_Debt",
    "Debt_to_Income_Ratio"
]])
```

```python
df["Loan_to_Income_Ratio"] = (
    df["Loan_Amount"] / df["Annual_Income"]
)

display(df[[
    "Loan_Amount",
    "Annual_Income",
    "Loan_to_Income_Ratio"
]])
```

```python
df["Credit_Utilization_Ratio"] = (
    df["Credit_Used"] /
    df["Credit_Limit"].replace(0, np.nan)
)

display(df[[
    "Credit_Used",
    "Credit_Limit",
    "Credit_Utilization_Ratio"
]])
```

```python
df["Employment_Stability"] = pd.cut(
    df["Employment_Years"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=[
        "Low_Stability",
        "Moderate_Stability",
        "High_Stability",
        "Very_High_Stability"
    ]
)

display(df[[
    "Employment_Years",
    "Employment_Stability"
]])
```

```python
df["Savings_to_Income_Ratio"] = (
    df["Monthly_Savings"] /
    (df["Annual_Income"] / 12)
)

display(df[[
    "Monthly_Savings",
    "Annual_Income",
    "Savings_to_Income_Ratio"
]])
```

```python
new_features = [
    "Debt_to_Income_Ratio",
    "Loan_to_Income_Ratio",
    "Credit_Utilization_Ratio",
    "Employment_Stability",
    "Savings_to_Income_Ratio"
]

print("First Five Domain-Based Features:")
display(df[["Customer_ID"] + new_features])

print("Total New Features Created:", len(new_features))
```

```python
df["Loan_Experience"] = pd.cut(
    df["Previous_Loans"],
    bins=[-1, 0, 2, 4, float("inf")],
    labels=["No_History", "Low", "Moderate", "High"]
)

display(df[["Previous_Loans", "Loan_Experience"]])
```

```python
df["Missed_Payment_Risk"] = pd.cut(
    df["Missed_Payments"],
    bins=[-1, 0, 2, float("inf")],
    labels=["Low", "Medium", "High"]
)

display(df[["Missed_Payments", "Missed_Payment_Risk"]])
```

```python
df["Available_Credit"] = (
    df["Credit_Limit"] - df["Credit_Used"]
)

display(df[[
    "Credit_Limit",
    "Credit_Used",
    "Available_Credit"
]])
```

```python
df["Disposable_Income"] = (
    (df["Annual_Income"] / 12)
    - df["Monthly_Debt"]
    - df["Monthly_Savings"]
)

display(df[[
    "Annual_Income",
    "Monthly_Debt",
    "Monthly_Savings",
    "Disposable_Income"
]])
```

```python
df["Financial_Risk_Score"] = (
    df["Debt_to_Income_Ratio"].fillna(0) * 40
    + df["Credit_Utilization_Ratio"].fillna(0) * 30
    + df["Missed_Payments"] * 10
)

display(df[[
    "Debt_to_Income_Ratio",
    "Credit_Utilization_Ratio",
    "Missed_Payments",
    "Financial_Risk_Score"
]])
```
