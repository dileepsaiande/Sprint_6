```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Income": [30000, 45000, 60000, 75000, 90000],
    "Age": [22, 28, 35, 42, 50],
    "Quantity": [2, 3, 4, 5, 6],
    "Price": [500, 700, 800, 900, 1000],
    "Experience": [1, 3, 8, 12, 20],
    "Total_Spending": [5000, 8000, 12000, 18000, 25000],
    "Number_of_Orders": [2, 4, 5, 6, 10]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
df["Quantity_Price"] = df["Quantity"] * df["Price"]

display(df[["Quantity", "Price", "Quantity_Price"]])
```

```python
df["Income_Experience_Sum"] = df["Income"] + df["Experience"]

display(df[["Income", "Experience", "Income_Experience_Sum"]])
```

```python
df["Income_Expense_Difference"] = df["Income"] - df["Total_Spending"]

display(df[["Income", "Total_Spending", "Income_Expense_Difference"]])
```

```python
df["Spending_Per_Order"] = (
    df["Total_Spending"] / df["Number_of_Orders"]
)

display(df[["Total_Spending", "Number_of_Orders", "Spending_Per_Order"]])
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2, include_bias=False)

poly_data = poly.fit_transform(df[["Age", "Income"]])

poly_df = pd.DataFrame(
    poly_data,
    columns=poly.get_feature_names_out(["Age", "Income"])
)

print("Polynomial Features:")
display(poly_df)
```

```python
from sklearn.preprocessing import PolynomialFeatures

interaction = PolynomialFeatures(
    degree=2,
    interaction_only=True,
    include_bias=False
)

interaction_data = interaction.fit_transform(
    df[["Income", "Age", "Experience"]]
)

interaction_df = pd.DataFrame(
    interaction_data,
    columns=interaction.get_feature_names_out(
        ["Income", "Age", "Experience"]
    )
)

print("Pairwise Interaction Features:")
display(interaction_df)
```

```python
df["Income_Per_Experience"] = (
    df["Income"] / df["Experience"].replace(0, np.nan)
)

df["Average_Order_Value"] = (
    df["Total_Spending"] / df["Number_of_Orders"]
)

df["Total_Purchase_Value"] = df["Quantity"] * df["Price"]

print("Domain-Based Interaction Features:")
display(df[[
    "Income_Per_Experience",
    "Average_Order_Value",
    "Total_Purchase_Value"
]])
```

```python
df["Income_Age"] = df["Income"] * df["Age"]

display(df[["Income", "Age", "Income_Age"]])
```

```python
df["Total_Price"] = df["Quantity"] * df["Price"]

display(df[["Quantity", "Price", "Total_Price"]])
```

```python
df["Experience_Age_Ratio"] = (
    df["Experience"] / df["Age"].replace(0, np.nan)
)

display(df[["Experience", "Age", "Experience_Age_Ratio"]])
```

```python
df["Spending_Per_Order"] = (
    df["Total_Spending"] /
    df["Number_of_Orders"].replace(0, np.nan)
)

display(df[[
    "Total_Spending",
    "Number_of_Orders",
    "Spending_Per_Order"
]])
```

```python
print("Final Dataset with Interaction Features:")
display(df)

print("Dataset Shape:", df.shape)
```
